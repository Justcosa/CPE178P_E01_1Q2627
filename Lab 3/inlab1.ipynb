```python
train = torchvision.datasets.ImageFolder("./flower_photos_train/", transform=transforms.ToTensor()) # Load the data as tensors.

# Check the dataset attributes.
print ("Dataset types")
print(train.classes)
print(train.class_to_idx)
print ("Dataset samples")
print(train.samples[0])
print ("Image data format")
print(train.extensions[0])
```

```python
# Display the loaded image data.
rng = random.Random(2)
for i in range(5):
    plt.figure()
    num = rng.randint(0, 3000)
    plt.imshow(train[num][0].permute(1,2,0)) # train[num] returns a tuple containing data and the category number corresponding to the label.
    plt.title(train[num][1])
    plt.colorbar()
    plt.grid(False)
plt.show()
```
