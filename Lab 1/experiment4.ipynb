```python
"""
If you are seeing this, please note that this code is modified. 
This is not the default code for Lab 1, experiment 4
"""

import pandas as pd
import numpy as np
from sklearn import tree
import pydotplus
```

```python
def createTree(trainingData, split_criterion="gini"):
    data = trainingData.iloc[:, :-1]  # Feature matrix
    labels = trainingData.iloc[:, -1]  # Labels
    
    # Set the decision tree classifier to use Gini impurity (or entropy if passed)
    trainedTree = tree.DecisionTreeClassifier(criterion=split_criterion)  
    trainedTree.fit(data, labels)  # Train the model.
    return trainedTree
```

```python
def showtree2pdf(trainedTree,finename):
    dot_data = tree.export_graphviz(trainedTree, out_file=None) # Export the tree in Graphviz format.
    graph = pydotplus.graph_from_dot_data(dot_data)
    graph.write_pdf(finename)  # Save the tree diagram to the local machine in PDF format.
```

```python
def data2vectoc(data):
    names = data.columns[:-1]
    for i in names:
        col = pd.Categorical(data[i])
        data[i] = col.codes
    return data
```

```python
# 1. Read training data.
data = pd.read_table("Dataset/ML/tennis.txt", header=None, sep='\t') 

# 2. Vectorize data.
trainingvec = data2vectoc(data) 

# 3. Create a decision tree using the Gini criterion
decisionTree = createTree(trainingvec, split_criterion="gini") 

# 4. Plot the decision tree and save it as a new PDF
showtree2pdf(decisionTree, "tennis_gini.pdf")
```

```python
# 5. Predict using the new model.
# (Weather is sunny, temperature is low, humidity is high, and wind is strong.)
testVec = [0, 0, 1, 1] 
prediction = decisionTree.predict(np.array(testVec).reshape(1, -1))
print("Prediction:", prediction)
```
